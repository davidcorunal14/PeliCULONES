# Pregunta 4 — Com ha canviat la durada de les pel·lícules (2006–2016)?
**Projecte:** Anàlisi d'un conjunt de dades (0378 – Seguretat i alta disponibilitat, Tema 1)
**Dataset:** `IMDB-Movie-Data.csv` — 1.000 pel·lícules (2006–2016)

Aquest notebook treballa **una única pregunta de recerca**, amb el procediment complet: preparació de les dades, mètode d'anàlisi, resultat i interpretació.

## 1. Preparació de les dades

**Procediment:**
- Carreguem el CSV en un `DataFrame` de Pandas.
- Renombrem columnes amb espais/parèntesis (`runtime`, `revenue`).
- Convertim `Revenue = 0` a `NaN` (no és una recaptació real, és un valor desconegut).
- Separem `genre` i `actors` en llistes (cada pel·lícula en pot tenir diversos).
- Creem un subconjunt `df_fiable` amb `votes >= 1000`, perquè les pel·lícules amb pocs vots tenen una nota poc representativa (sobretot les de 2016).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

df = pd.read_csv("IMDB-Movie-Data.csv", encoding="utf-8")
df = df.rename(columns={"Runtime (Minutes)": "runtime", "Revenue (Millions)": "revenue"})
df.columns = [c.lower() for c in df.columns]

df.loc[df["revenue"] == 0, "revenue"] = np.nan

df["genres"] = df["genre"].str.split(",").apply(lambda l: [g.strip() for g in l])
df["actors_list"] = df["actors"].str.split(",").apply(lambda l: [a.strip() for a in l])

df_fiable = df[df["votes"] >= 1000].copy()

print(f"Total de pel·lícules: {len(df)}")
print(f"Amb >= 1000 vots (notes fiables): {len(df_fiable)}")
df.head(3)
```

## 2. Pregunta de recerca

**Com ha canviat la durada mitjana de les pel·lícules entre 2006 i 2016?**

**Mètode:** agrupem per `year` i calculem la mitjana, la mediana i el nombre de pel·lícules de `runtime` per a cada any. Afegim el nombre de pel·lícules per any perquè el dataset no té el mateix volum tots els anys (2016 en té moltes més que 2006), cosa que cal tenir en compte en interpretar la tendència.

```python
resum_any = df.groupby("year")["runtime"].agg(mitjana="mean", mediana="median", n="count").round(1)
resum_any
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
resum_any["mitjana"].plot(marker="o", ax=ax, label="Mitjana")
resum_any["mediana"].plot(marker="s", ax=ax, label="Mediana", linestyle="--")
ax.set_title("Durada de les pel·lícules per any")
ax.set_xlabel("Any"); ax.set_ylabel("Minuts"); ax.legend()
plt.tight_layout(); plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
resum_any["n"].plot(kind="bar", ax=ax, color="lightgrey")
ax.set_title("Nombre de pel·lícules del dataset per any")
plt.tight_layout(); plt.show()
```

## 3. Resultats

- La durada mitjana baixa de manera clara: de **120,8 min el 2006** a **107,4 min el 2016**, uns 13 minuts menys en una dècada.
- La baixada no és constant: hi ha un petit repunt el 2012 (119,1 min) abans de tornar a baixar.
- **Important:** el 2016 té 297 pel·lícules al dataset (gairebé el 30% del total), mentre que 2006 només en té 44. La xifra de 2016 pesa molt en qualsevol mitjana global, i part de la baixada podria estar relacionada amb quines pel·lícules de 2016 s'han inclòs al dataset, no necessàriament amb un canvi real a la indústria.


## 4. Interpretació i limitacions

✏️ **Omple-ho tu:** La baixada sembla real, però cal anar amb compte: 2016 és el 30% del dataset i pesa molt en la mitjana. Podria tenir relació amb l’auge de l’streaming, però amb aquestes dades no es pot assegurar.
